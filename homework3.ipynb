```python
def letter_score(letter):
    if letter in ['a', 'e', 'i', 'l', 'n', 'o', 'r', 's', 't', 'u']:
        return 1
    elif letter in ['d', 'g']:
        return 2
    elif letter in ['b', 'c', 'm', 'p']:
        return 3
    elif letter in ['f', 'h', 'v', 'w', 'y']:
        return 4
    elif letter in ['k']:
        return 5
    elif letter in ['j', 'x']:
        return 8
    elif letter in ['q', 'z']:
        return 10
    else:
        return 0

print('letter_score("j") returned', letter_score('j'))   
print('letter_score("k") returned', letter_score('k'))   
print('letter_score("e") returned', letter_score('e'))   
```

Problem 6

Concept: Decision logic

```python
def calculate_weekly_pay(hours_worked, rate):
    weekly_pay = 0

    if hours_worked <= 40:
        weekly_pay = hours_worked * rate
    elif hours_worked > 40 and hours_worked <= 55:
        weekly_pay = (40 * rate) + ((hours_worked - 40) * (rate * 1.5))
    else:
        weekly_pay = (40 * rate) + ((15) * (rate * 1.5)) + ((hours_worked - 55) * (rate * 2))
    return weekly_pay

# test case
print('calculate_weekly_pay(25, 20)', calculate_weekly_pay(25, 20))
# test case
print('calculate_weekly_pay(45, 20)', calculate_weekly_pay(45, 20))
# test case
print('calculate_weekly_pay(60, 20)', calculate_weekly_pay(60, 20)) 
```

Problem 7

Concept: Decision logic.

```python
def longest_string(strings):
  if len(strings[0]) > len(strings[1]) and len(strings[0]) > len(strings[2]):
    return strings[0]
  elif len(strings[1]) > len(strings[0]) and len(strings[1]) > len(strings[2]):
    return strings[1]
  else:
    return strings[2]

# test case
print('longest_string(["all", "my", "loving"]) returned', longest_string(["all", "my", "loving"]))
print('longest_string(["strawberry", "fields", "forever"]) returned', longest_string(["strawberry", "fields", "forever"]))
```

Problem 8

Concept: Decision logic with strings.

```python
def alphabetize(s1, s2, s3):
  if s1 < s2 and s1 < s3:
    if s2 < s3:
      return [s1, s2, s3]
    else:
      return [s1, s3, s2]
  elif s2 < s1 and s2 < s3:
    if s1 < s3:
      return [s2, s1, s3]
    else:
      return [s2, s3, s1]
  else:
    if s1 < s2:
      return [s3, s1, s2]
    else:
      return [s3, s2, s1]

# test case
print('alphabetize("all", "my", "loving") returned', alphabetize("all", "my", "loving"))

print('alphabetize("strawberry", "fields", "forever") returned', alphabetize("strawberry", "fields", "forever"))

print('alphabetize("zebra", "apple", "monkey") returned', alphabetize("zebra", "apple", "monkey"))

print('alphabetize("money", "apple", "monkey") returned', alphabetize("money", "apple", "monkey"))
```

Problem 13

Concept: List comprehension with aggregate function.

```python
def stdev(values):
    import math
    mean = sum(values) / len(values)
    varience_values = [(x - mean) ** 2 for x in values]
    variance = sum(varience_values) / len(values)
    return math.sqrt(variance)

# test case

print('stdev([1, 2, 3, 4, 5]) returned', stdev([1, 2, 3, 4, 5]))
```

Problem 14

Concept: List comprehension, using helper function.

```python
def scrabble_score(word):
    letter_scores = [letter_score(letter) for letter in word]
    return sum(letter_scores)

#test case
print('scrabble_score("strawberry") returned', scrabble_score("strawberry"))
print('scrabble_score("pie") returned', scrabble_score("pie"))
```

Problem 15

Concept: List comprehension, using helper functions.

```python
def scrabble_score_list(words):
    return [scrabble_score(word) for word in words]

#test case
print('scrabble_score_list(["a", "b", "c"]) returned', scrabble_score_list(["a", "b", "c"]))
print('scrabble_score_list(["strawberry", "fields", "forever"]) returned', scrabble_score_list(["strawberry", "fields", "forever"]))
```
