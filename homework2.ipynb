Task:

Write two functions with the exact names below:

```first_and_last(numbers)```

Returns the sum of the first and last elements in the list numbers.

```mean_of_list(numbers)```

Returns the mean (average) of all values in the list numbers.

Use the provided test cases.

````python
def first_and_last(numbers):
    """Return the sum of the first and last elements of a list of numbers"""

    return numbers[0] + numbers[-1]

def mean_of_list(numbers):
    """Return the mean of a list of numbers"""

    return sum(numbers) / len(numbers)

# test cases for your functions
print('first_and_last([6,1,7,3,5,8,2,3,6,1]) =', first_and_last([6,1,7,3,5,8,2,3,6,1]))
print('mean_of_list([6,1,7,3,5,8,2,3,6,1]) =', mean_of_list([6,1,7,3,5,8,2,3,6,1]))
````

Problem 8

Concept: Function with a list as parameter.

Task:

Write a function named ```polynomial(x, coefficients)``` that:

Takes a number ```x```

Takes a list named ```coefficients``` containing exactly four values

Computes the polynomial:

```ax^3 + bx^2 + cx + d```

Use list indexing to access the coefficients.

````python
def polynomial(x, coefficients):
    """Return the value of a polynomial given x and a list of coefficients"""

    result = 0
    for i in range(len(coefficients)):
        result += coefficients[i] * (x ** (len(coefficients) - 1 - i))
    return result

# test cases for your function:
print('polynomial(1, [2, 3, 4, 5]) =', polynomial(1, [2, 3, 4, 5]))
print('polynomial(2, [2, 3, 4, 6]) =', polynomial(2, [2, 3, 4, 6]))
````

Problem 9

Concept: Function with a parameter which is a list of strings.

Task:

Write a function named ```make_initials(names)``` that:

Takes a list named ```names``` containing exactly three strings (first, middle, last)

Returns a string consisting of the first character of each name

````python
def make_initials(names):
    """Return a list of initials given a list of names"""

    string = ""
    for name in names:
        string += name[0]
    return string


# test cases for your function
print('make_initials(["john", "winston", "lennon"]) =', make_initials(["john", "winston", "lennon"]))
print('make_initials(["james", "paul", "mccartney"]) =', make_initials(["james", "paul", "mccartney"]))
````

Task:

Write a function named ```first_mid_last(my_list)``` that:

Takes a list named ```my_list```

Returns a new list containing:

First element

Middle element

Last element

Note:

For lists with even number of elements, you should return the second middle element. See sample test cases for details.

````python
def first_mid_last(my_list):
    """Return a list containing the first, middle, and last elements of a list"""

    return [my_list[0], my_list[len(my_list) // 2], my_list[-1]]

# test cases for your function

print('first_mid_last([1, 2, 3, 4, 5]) =', first_mid_last([1, 2, 3, 4, 5]))
print('first_mid_last([6,1,7,3,5,8,2,3,6,1]) =', first_mid_last([6,1,7,3,5,8,2,3,6,1]))
````

Problem 11

Concept: Function that returns a list.

Task:

Write a function named ```change_from_a_dollar(price)``` that:

Takes an integer ```price``` (in cents, less than 100)

Calculates how much change is needed from one dollar (100 cents)

Returns a list in the following order:

Number of quarters (1 quarter = 25 cents)

Number of dimes (1 dime = 10 cents)

Number of nickels (1 nickel = 5 cents)

Number of pennies (1 penny = 1 cent)

````python
def change_from_a_dollar(price):
    """Return the change from a dollar given a price"""

    change = 100 - price
    quarter = 25
    dime = 10
    nickel = 5
    penny = 1

    number_of_quarters = int(change / quarter)
    number_of_dimes = int((change - (number_of_quarters * quarter)) / dime)
    number_of_nickels = int((change - (number_of_quarters * quarter) - (number_of_dimes * dime)) / nickel)
    number_of_pennies = int((change - (number_of_quarters * quarter) - (number_of_dimes * dime) - (number_of_nickels * nickel)) / penny)
    list_of_change = [number_of_quarters, number_of_dimes, number_of_nickels, number_of_pennies]
    return list_of_change

# test cases for your function
print("change_from_a_dollar(22) returned", change_from_a_dollar(22))
print("change_from_a_dollar(42) returned", change_from_a_dollar(42))
````
