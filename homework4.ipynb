Task:

Write a function middlest_element(values) that returns the middle element of a list.

Rules:

If the list is empty, return None.

If the list length is odd, return the element at the exact middle index.

If the list length is even, return the element just left of center (example: for length 6, use index 2).

Then run the provided test cases and print results exactly as shown.

```python
def middlest_element(values): 
    list_lenght =len(values)
    if list_lenght == 0:
        return None
        
    middle_index=(len(values)-1)//2  

    return values[middle_index]

# test case
values = [0,1,2,3,7,8,9]
print('middlest_element(values) = ', middlest_element(values))

values = [1,2,3,7,8,9]
print('middlest_element(values) = ', middlest_element(values))

values = []
print('middlest_element(values) = ', middlest_element(values))
```

Concept: List comprehensions.

Task:

Write a function square_evens(values) that returns a new list containing the square of every even number in values, in the same order they appear.

Rules:

Must use a list comprehension.

Then write your own test cases that clearly prove correctness (include at least: mixed evens/odds, all odds, empty list, negative evens).

```python
def square_evens(values):
    square_evens_list=[]
    for num in values:
        if num % 2==0:
            square_evens_list = square_evens_list + [num * num]
    return square_evens_list

square_evens([3, 4, 2, 6, 9, 8])
```

Write a function print_powers(base, n) that prints the first n+1 powers of base, starting from exponent 0 up to exponent n.

```python
def print_powers(base, n):
    print("printing the first %d powers of %d" % (n+1,base))
    for i in range(n+1):
        print("%d ** %d = %d" % (base,i,pow(base,i)))

print_powers(2, 10)
```

Rules:

Concept: Accumulator pattern with numeric result, list iteration.

Task:

Write a function variance(values) that returns the population variance of a list of numbers using this formula:

 

Rules:

Must use loops + accumulator (no sum()).

No imports.

Must call your mean(values) function to get the mean.

Then run the provided test case and print in the same format.

```python
def variance(values):    
    before_variance=0.0
    mean_value=mean(values)
    for num in values:
         before_variance= before_variance + pow(num - mean_value,2)
    return before_variance/len(values)


values_variance=[6, 3, 7, 9, 5]
print(f"values: {values_variance}")
print(f"mean: {mean(values_variance)}") 
print(f"variance: {variance(values_variance)}")  
```

Problem 8

Concept: Accumulator pattern with numeric result, list iteration, index-based loop.

Task:

Write a function covariance(xs, ys) that computes the population covariance between two series:

 

Rules:

If xs and ys are not the same length, print:

xs and ys must be the same length

and return 0.

Must use an index-based loop.

Must use your mean() function.

Then run the given test case and print exactly in the same format.

```python
def covariance(xs, ys):
    if(len(xs)!= len(ys)):
        print("xs and ys must be the same length")
        return 0
    before_covariance=0
    x_mean = mean(xs)
    y_mean = mean(ys)
    for i in range(len(xs)):
        before_covariance = before_covariance + ((xs[i]-x_mean)*(ys[i]-y_mean))
    return before_covariance/len(xs)

x_values=[6, 3, 7, 9, 5]
y_values =[4, 1, 7, 5, 0]
print(f"x values: {x_values}")
print(f"y values: {y_values}")
print(f"covariance: {covariance(x_values,y_values)}") 
```

Must loop through the string characters.

```python
def string_to_ascii(s):
    ascii_code =[]
    for ch in s:
        ascii_code=ascii_code+[ord(ch)]
    return ascii_code

# test cases
result = string_to_ascii("hello")
print(f"result = {result}")
```

Problem 12

Concept: Accumulator pattern to process a string, produce a string result.

Task:

Write a function cipher(s, n) that applies a Caesar cipher shift to a plaintext string.

Rules:

Assume lowercase letters only for shifting (a to z).

Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

Must wrap around the alphabet if shifting past z or before a.

Must use an accumulator string.

Then run the exact test cases shown and print in the same format.

```python
def cipher(s, n):
    result=""
    for ch in s:
        if (ch.isalpha() and ch.islower()):
            position = ord(ch)-ord('a')
            new_position = position+n
            wrapped=new_position%26
            result=result+chr(wrapped+ord('a'))
        else:
            result=result+ ch

    return result

# test cases:

# simple shift
plaintext = "abc"
result = cipher(plaintext, 3)
print(f"plaintext = {plaintext}, ciphered = {result}")

# wrap around, non-letter:
plaintext = "abc xyz"
result = cipher(plaintext, 3)
print(f"plaintext = {plaintext}, ciphered = {result}")
```
